This problem generator has been written to help with practicing applying Gauss-Seidel and Jacobi iteration. It generates a random matrix and vector, and then solves the system of equations using both methods. The results are then compared to the actual solution to determine the error.


```python
import numpy as np
np.seterr(all="raise")
```

```python
def jacobi(A, x0, b, iterations):
    x = x0.copy()
    n = len(x)
    
    # store the iterations
    x_iter = [x]
    
    for it in range(iterations):
        xnew = x.copy()
        
        for i in range(n):
            # form residual component i
            res_i = b[i].copy()
            for j in range(n):
                res_i -= A[i, j] * x[j]
                
            # apply iteration
            xnew[i] = x[i] + 1/A[i, i] * res_i
            
            x_iter.append(xnew)
            x = xnew.copy()
        
    return x_iter

def gauss_seidel(A, x0, b, iterations):
    x = x0.copy()
    n = len(x)
    
    # store the iterations
    x_iter = [x]
    
    for it in range(iterations):
        xnew = x.copy()
        
        for i in range(n):
            # form residual component i
            res_i = b[i].copy()
            for j in range(i):
                res_i -= A[i, j] * xnew[j]
            for j in range(i+1, n):
                res_i -= A[i, j] * x[j]
                
            # apply iteration
            xnew[i] = x[i] + 1/A[i, i] * res_i
            
            x_iter.append(xnew)
            x = xnew.copy()
            
    return x_iter

def latex_array(v):
    text = "\\begin{pmatrix}" + ' \\\\ '.join([' & '.join([str(coefficient) for coefficient in row]) for row in v]) + "\\end{pmatrix}"
    return text

def problem_generator(matrix_size, iterations, max_value=5):
    # loop until we get a good problem
    while True:
        # generator random problem
        A = np.random.randint(-max_value, max_value, size=(matrix_size, matrix_size))
        b = np.random.randint(-max_value, max_value, size=(matrix_size, 1))
        x0 = np.random.randint(-max_value, max_value, size=(matrix_size, 1))
        
        # get solutions and check output
        try:
            x_iter_jacobi = jacobi(A, x0, b, iterations)
            x_iter_gauss_seidel = gauss_seidel(A, x0, b, iterations)
                        
            break
        except:
            continue


    # convert problem to text
    A_text = latex_array(A)
    x_text = "\\begin{pmatrix}" + ' \\\\ '.join([f"x_{j+1}" for j in range(matrix_size)]) + "\\end{pmatrix}"
    b_text = latex_array(b)
    x0_text = latex_array(x0)
    x_jacobi_text = latex_array(x_iter_jacobi[-1])
    x_gauss_seidel_text = latex_array(x_iter_gauss_seidel[-1])
    
    # write problem text
    q = f"Consider the system of linear equations $${A_text} {x_text} = {b_text}.$$\n\n"
    
    q += f" - Apply {iterations} steps of Jacobi iteration starting from $\\vec{{x}}^{{(0)}} = {x0_text}$."
    q += f"Show that $\\vec{{x}}^{{({iterations})}} = {x_jacobi_text}$.\n"
    
    q += f" - Apply {iterations} steps of Gauss-Seidel iteration starting from $\\vec{{x}}^{{(0)}} = {x0_text}$."
    q += f"Show that $\\vec{{x}}^{{({iterations})}} = {x_gauss_seidel_text}$.\n"
    
    return q    
```
